```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#       0          1x       2x        3x        4x        5x         6x
base = [17682.54, 17212.81, 15894.11, 18753.36, 17423.54, 15499.88,  17796.75]
sumit = [18753.36, 12310.04, 7783.21, 8019.43, 8983.71, 7130.23, 8674.55]
sumit_parsa = [6191.11, 5768.87, 5751.65, 5539.46, 5216.39, 5203.89, 6029.59]
```

```python
plt.plot(base, label='Baseline')
plt.plot(sumit, label='Length Dist')
plt.plot(sumit_parsa, label='Length Dist + Pruning')
plt.ylim(0, 20_000)
plt.xlabel('Epoch')
plt.ylabel('Time (s)')
plt.title('Total Generation Time for Inference Methods')
plt.grid(True)
plt.legend()
plt.show()
```

```python
import pandas as pd
import seaborn as sns

df = pd.DataFrame({
    'Epoch': list(range(len(base))) * 3,
    'Time (s)': list(base) + list(sumit) + list(sumit_parsa),
    'Method': ['Baseline'] * len(base) + ['Length Dist'] * len(sumit) + ['Length Dist + Pruning'] * len(sumit_parsa),
})

plt.figure(figsize=(10, 6))
sns.lineplot(data=df, x='Epoch', y='Time (s)', hue='Method', marker='o')
plt.title('Total Generation Time for Inference Methods')
plt.ylabel('Time (s)')
plt.xlabel('Epoch')
plt.ylim(0, 20_000)
plt.grid(True)
plt.show()
```
